```python
# Strings (Concatenation, Slicing, and Other Methods) :
# 1. String Concatenation:

string1 = "Hello"
string2 = input("Enter your name:")
a=string1 +" "+ string2
print(a)
string3 = "Welcome to Python programming"
print(a +", "+ string3)
```

```python
# 2. String Slicing and Indexing:

# Get above concatenated string as input
text = input("Enter the String: ")

# a. Print the first character of the string
print("Print the first character of the string:",text[0])

# b. Print the last character of the string.
print("Print the last character of the string:",text[-1])

# c. Print the first 5 characters of the string.
print("Print the first 5 characters of the string:",text[:6])

# d. Print the last 11 characters of the string.
print("Print the last 11 characters of the string:",text[-11:])

# e. Print the string in reverse.
print("Print the string in reverse:",text[::-1])

# f. Use slicing and print the word “Python” from the existing string
# skills=data.split(",")
# print(skills)

B = text.split(" ")
print(B)
print(B[4])

# OR
C = "Hello Zara, Welcome to Python programming"
start = C.index("Python")
print(C[start:start + 6])
```

```python
# 3. String Methods:

strM = "Python beginner tutorial"
print(f'strM=',strM)

# a. Convert the sentence to uppercase.
print(f'Changed upper case=',strM.upper())

# b. Convert the sentence to lowercase.
print(f'Changed lower case=',strM.lower())

# c. Use Capitalize and return the sentence to the original input form.
print(f'Changed Capitalize=',strM.capitalize())

# d. Count the total number of occurrences of character ‘t’ in the string.
print(f'Count of "t" charecters=',strM.count("t"))

# e. Replace all occurrences of “Python” with “Data Analytics” in the input string strM = “Python beginner tutorial.”
strM = strM.replace('Python','Data Analytics')
print(f'Replace string=',strM)
```

```python
# 4. Tuples (Creation, Modification and Access) :

tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)
print(tuple1)
print(tuple2)

# a. Concatenate the two tuples and store it in “t_combine”

t_combine = tuple1 + tuple2
print(t_combine)


# b. Repeat the elements of “t_combine” 3 times

print(t_combine*3)

# c. Access the 3rd element from “t_combine”

print(t_combine[2])

# d. Access the first three elements from “t_combine”

print(t_combine[0:3])

# e. Access the last three elements from “t_combine”

print(t_combine[-3:])
```
